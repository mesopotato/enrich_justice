```python
import requests

url = "http://localhost:5010/api/parseDocument?renderFormat=all" #&useNewIndentParser=yes&strategy=sections&strategy=sections
files = {'file': open(r'Y:\entscheidsuche\BE_Anwaltsaufsicht\PDF\BE_AK_001_AA-2012-98_2014-09-02.pdf', 'rb')}
response = requests.post(url, files=files)
print(response.json())
```

```python
import json
# Pretty-print the JSON response
print(json.dumps(response.json(), indent=4))
```

```python
def print_structure(data, indent=0):
    if isinstance(data, dict):
        for key, value in data.items():
            print(' ' * indent + f"{key}: {type(value).__name__}")
            print_structure(value, indent + 2)
    elif isinstance(data, list):
        print(' ' * indent + f"List of {len(data)} items")
        if len(data) > 0:
            print_structure(data[0], indent + 2)

# Print the high-level structure of the JSON response
print_structure(response.json())
```

```python
from llmsherpa.readers import LayoutPDFReader
# adding "&useNewIndentParser=yes&strategy=sections" does not change anything!
llmsherpa_api_url = "http://localhost:5010/api/parseDocument?renderFormat=all"
pdf_url = "https://entscheidsuche.ch/docs/BE_AK_001_AA-2012-98_2014-09-02.pdf"
#https://entscheidsuche.ch/docs/BE_ZivilStraf/BE_OG_001_KES-2013-542_2013-12-16.pdf
pdf_reader = LayoutPDFReader(llmsherpa_api_url)
doc = pdf_reader.read_pdf(pdf_url)
```

```python
for chunk in doc.chunks():
    print(chunk.to_text())
```

```python
for table in doc.tables():
    print(table.to_text())
```

```python
for section in doc.sections():
    print(section.to_text())
```

```python
print(doc.to_text())
```
